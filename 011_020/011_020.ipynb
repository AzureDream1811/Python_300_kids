## 011 - Write a function to calculate the factorial of a number

```python
def factorial(n: int):
    return 1 if n == 0 else n * factorial(n - 1)

def factorial_iterative(n: int):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial(5))
print(factorial_iterative(5))
```

## 012 - Write a function to calculate the sum of numbers from 1 to n

```python
def sum_of_nums(n: int):
    return 1 if n == 1 else n + sum_of_nums(n - 1)

print(sum_of_nums(8)) # 1 + 2 + 3 + 4 + 5 + 6 + 7 + 8 = 36
```

## 013 - Write a function to check if a number is a prime number.

```python
def is_prime(n: int):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i**2 <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True
    
print(is_prime(11))
```

## 014 - Write a function to find the largest number in a list

```python
def largest_in_list(given_list: list):
    largest = 0
    for num in given_list:
        largest = max(num, largest)
    return largest

given_list = [5, 10, 15, 20, 11, 13, 15, 21, 13, 9]

print(largest_in_list(given_list))
```

## 015 - Write a function to calculate the sum of the digits of a number.

```python
def sum_of_digits(number: int):
    number = abs(number)
    total = 0
    while number > 0:
        total += number % 10
        number //= 10
    return total

print(sum_of_digits(561))    
```

## 016 - Write a function to check if a string is a palindrome

```python
def is_palindrome(string: str):
    string = ''.join(char.lower() for char in string if char.isalnum())
    return string == string[::1]

print(is_palindrome('1s21@12s1'))
```

## 017 - Write a function to calculate the power of a number.

```python
def power(base, exponent):
    result = 1
    if (exponent == 0):
        return result


    for _ in range(exponent):
        result *= base

    return result

print(power(2, 0))
print(power(5, 2))
print(power(7, 4))
```

## 018 - Write a function to convert degrees Celsius to degrees Fahrenheit

```python
def convert_degrees(degrees, to_celsius=True):
    if to_celsius is True:
        return (degrees - 32) * 5 / 9
    else:
        return degrees * 9 / 5 + 32


print(f"{round(convert_degrees(100), 2)} C")
print(f"{round(convert_degrees(10, False), 2)} F")
```

## 019 - Write a function to calculate the circumference and area of ​​a circle

```python
import math


def calc_circumference_cicrle(radius):
    return 2 * radius * math.pi


def calc_area_cicle(radius):
    return power(radius, 2) * math.pi


radius = 5
print(f"Radius = {radius}")
print(f"Circumference: {round(calc_circumference_cicrle(radius), 2)}")
print(f"Area: {round(calc_area_cicle(radius), 2)}")
```

## 020 - Write a function to calculate the perimeter and area of ​​a rectangle

```python
def calc_circumference_rec(width, height):
    return (width + height) * 2


def calc_area_rec(width, height):
    return width * height


width = 8
height = 5
print(f"Width = {width}")
print(f"Height = {height}")
print(f"Circumference: {round(calc_circumference_rec(width, height), 2)}")
print(f"Area: {round(calc_area_rec(width, height), 2)}")
```
