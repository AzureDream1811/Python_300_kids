## 011 - Write a function to calculate the factorial of a number

```python
def factorial(n: int):
    return 1 if n == 0 else n * factorial(n - 1)

def factorial_iterative(n: int):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

print(factorial(5))
print(factorial_iterative(5))
```

## 012 - Write a function to calculate the sum of numbers from 1 to n

```python
def sum_of_nums(n: int):
    return 1 if n == 1 else n + sum_of_nums(n - 1)

print(sum_of_nums(8)) # 1 + 2 + 3 + 4 + 5 + 6 + 7 + 8 = 36
```
