## 037 - Viết chương trình để kiểm tra một phần tử có tồn tại trong tuple không


```python
def is_ele_in_tuple(element, tuple_):
    return element in tuple_

print(is_ele_in_tuple(2, tuple1))
```

## 038 - Viết chương trình để đếm số lần xuất hiện của một phần tử trong tuple


```python
def count_each_element_in_tuple(tuple_):
    dict_count = {}
    for element in tuple_:
        if element in dict_count:
            dict_count[element] += 1
        else:
            dict_count[element] = 1

    return dict_count

print(count_each_element_in_tuple(tuple3))
```

## 039/040 - Viết chương trình để tìm phần tử lớn/nhỏ nhất trong tuple


```python
def find_min_max_in_tuple(tuple_):
    return min(tuple_), max(tuple_)

print(find_min_max_in_tuple(tuple3))
```
