## 021 - Write a program to create a list of numbers from 1 to 100

```python
def create_list(end):
    return list(range(1, end))

numbers = create_list(101)
print(numbers)
```

## 022 - Write a program to add an element to the end of a list

```python
def add_end_list(list, ele):
    list.append(ele)
    return list

print(add_end_list(numbers, 15))
```

## 023 - Write a program to add an element to the beginning of a list

```python
def add_start_list(list, ele):
    list.insert(0, ele)
    return list

print(add_start_list(numbers, 0))
```

## 024 - Write a program to remove an element from a list

```python
def remove_element(list, ele):
    list.remove(ele)
    return list

print(remove_element(numbers, 15))
```

## 025 - Write a program to sort a list in ascending order

```python
def sort_list(list):
    return sorted(list, reverse=False)

print(sort_list(numbers))
```

## 026 - Write a program to sort a list in descending order

```python
print(sorted(numbers, reverse=True))
```

## 027 - Write a program to find the largest element in a list

```python
def find_max_min(list):
    return max(list), min(list)

print(find_max_min(numbers))
```

## 028 - Write a program to find the smallest element in a list

```python
print(find_max_min(numbers))
```

## 029 - Write a program to calculate the sum of the elements in a list

```python
def calc_sum(list):
    return sum(list)

print(calc_sum(numbers))
```

## 030 - Write a program to count the number of occurrences of an element in a list

```python
def count_each_element(list):
    return {i: list.count(i) for i in list}

print(count_each_element(numbers))
```
