## 081 - Viết chương trình để tạo một lớp học sinh


```python
class Student:
    def __init__(self, name, age, avg):
        self.name = name
        self.age = age
        self.avg = avg

    def view_info(self):
        print(f"Student: {self.name}")
        print(f"Age: {self.age}")
        print(f"Avg: {self.avg}")

    def rank(self):
        if self.avg >= 8.5:
            return "Excellent"
        elif self.avg >= 6.5:
            return "Good"
        elif self.avg >= 5.0:
            return "Average"
        else:
            return "Weak"


student = Student("Azure Dream", 22, 7.8)

student.view_info()

print(f"Rank: {student.rank()}")
```

## 085 - Viết chương trình để truy cập thuộc tính của đối tượng

```python
class Student:
    def __init__(self, name, age, avg, address):
        self.name = name
        self._age = age
        self.avg = avg
        self.__address = address

    def view_info(self):
        print(f"Student: {self.name}")
        print(f"Age: {self.age}")
        print(f"Avg: {self.avg}")

    def rank(self):
        if self.avg >= 8.5:
            return "Excellent"
        elif self.avg >= 6.5:
            return "Good"
        elif self.avg >= 5.0:
            return "Average"
        else:
            return "Weak"

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, new_age):
        if new_age > 0:
            self._age = new_age
        else:
            print("Input age invalid")

    def get_address(self):
        return self.__address

    def set_address(self, new_address):
        self.__address = new_address


student = Student("Azure Dream", 22, 7.8, "123/A Street, B city")

# Access and change public attributes
print(f"Before: Name - {student.name}, Average - {student.avg}")
student.name = "Faragoth"
student.avg = 9.0
print(f"After: Name - {student.name}, Average - {student.avg}")

# Access and change private attributes with getter và setter
print(f"Before: Age - {student.age}, Địa chỉ - {student.get_address()}")
student.age = 17
student.set_address("456/B Street, C city")
print(f"After: Age - {student.age}, Địa chỉ - {student.get_address()}")

student.view_info()
```

## 087 - Viết chương trình để kế thừa từ một lớp cơ bản


```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def view_info(self):
        print(f"Student: {self.name}")
        print(f"Age: {self.age}")

class Student(Human):
    def __init__(self, name, age, avg, address):
        super().__init__(name, age)
        self.avg = avg
        self.address = address

    def view_info(self):
        super().view_info()
        print(f"Avg: {self.avg}")
        print(f"Address: {self.address}")

    def rank(self):
        if self.avg >= 8.5:
            return "Excellent"
        elif self.avg >= 6.5:
            return "Good"
        elif self.avg >= 5.0:
            return "Average"
        else:
            return "Weak"

student = Student("Azure Dream", 22, 7.8, "123/A Street, B city")

student.view_info()
print(f'Rank: {student.rank()}')
```

## 089 - Viết chương trình để sử dụng trừu tượng trong Python


```python
from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):
    def sound(self):
        return "Woof!"


class Cat(Animal):
    def sound(self):
        return "Meow!"


dog = Dog()
cat = Cat()

print(dog.sound())
print(cat.sound())
```

## 090 - Viết chương trình để sử dụng đa hình trong Python


```python
class Animal:
    def sound(self):
        return "Animal makes a sound"

class Dog(Animal):
    def sound(self):
        return "Woof!"

class Cat(Animal):
    def sound(self):
        return super().sound()

animals = [Dog(), Cat()]

for animal in animals:
    print(animal.sound())
```
