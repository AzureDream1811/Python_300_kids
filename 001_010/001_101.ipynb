# 001: CHECK NUMBER

```python
def check_number(n):
    if n > 0:
        return 'The number is positive.'
    return 'The number is negative' if n < 0 else 'The number is zero'
        
print(check_number(10))
print(check_number(-10))
print(check_number(0))
```

# 002: ODD OR EVEN

```python
def check_even_odd(n : int):
        if n == 0:
            return 'The number is zero (Cannot odd or even)'
        return 'The number is even' if n % 2 == 0 else 'The number is odd'
    
print(check_even_odd(5))
print(check_even_odd(10))
print(check_even_odd(0))
```
