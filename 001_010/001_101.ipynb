# 001: CHECK NUMBER

```python
def check_number(n):
    if n > 0:
        return 'The number is positive.'
    return 'The number is negative' if n < 0 else 'The number is zero'
        
print(check_number(10))
print(check_number(-10))
print(check_number(0))
```

# 002: ODD OR EVEN

```python
def check_even_odd(n : int):
        if n == 0:
            return 'The number is zero (Cannot odd or even)'
        return 'The number is even' if n % 2 == 0 else 'The number is odd'
    
print(check_even_odd(5))
print(check_even_odd(10))
print(check_even_odd(0))
```

# 003: LARGEST OF THE THREE

```python
def find_largest(a, b, c):
    largest = a
    largest = max(largest, b)
    largest = max(largest, c)
    return largest

print(find_largest(10, 5, 15))
```

# 004: CALCULATE TAXI FEE

```python
def calc_taxi_fee(n: float):
    if n <= 1:
        return n * 10000
    if n <= 10:
        return 10000 + (n - 1) * 8500
    else:
        return 10000 + 9 * 8500 + (n - 10) * 7500


print(calc_taxi_fee(1))
print(calc_taxi_fee(4))  # 10000 + 3 * 8500 = 35500
print(calc_taxi_fee(13))  # 10000 + 9 * 8500 + 3 * 7500 = 109000
```
