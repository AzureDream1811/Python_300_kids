```python
import os

BASE = os.getcwd()

print(f"Current working directory: {BASE}")
```

## 071 - Viết chương trình để đọc nội dung của một file văn bản


```python
file_path = BASE + "\\example.txt"

try:
    with open(file_path, "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print(f"Error: '{file_path}' was not found.")
```

## 072 - Viết chương trình để ghi nội dung vào một file văn bản

```python
file_path = BASE + "\\example_output.txt"

content_to_write = "Hello world...."

try:
    with open(file_path, "w") as file:
        file.write(content_to_write)
        print(f"Content write into file: {file_path}")
except FileNotFoundError as e:
    print(f"Error: {e}")
```

## 074 - Viết chương trình để đếm số từ trong một file văn bản


```python
file_path = BASE + "\\example.txt"

try:
    with open(file_path, "r") as file:
        word_count = 0
        for line in file:
            print(line)
            words = line.split()
            word_count += len(words)

        print(f"Number of words: {word_count}")
except FileNotFoundError as e:
    print(f"Erro: {e}")
```

## 075 - Viết chương trình để đếm số ký tự trong một file văn bản

```python
file_path = BASE + "\\example.txt"

try:
    with open(file_path, "r") as file:
        content = file.read()
        filtered_content = ''.join(c for c in content if c not in (' ', '\n', '\t'))
        char_count = len(filtered_content)
        print(f'Number of char: {char_count}')
except FileNotFoundError as e:
    print(f"Erro: {e}")
```

## 076 - Viết chương trình để kiểm tra một file có tồn tại không

```python
import os

file_path = 'example.txt'

if os.path.isfile(file_path):
    print(f"File '{file_path}' tồn tại.")
else:
    print(f"File '{file_path}' không tồn tại.")
```

## 077 - Viết chương trình để xóa một file văn bản


```python
import os

file_path = 'example.txt'

try:
    if os.path.isfile(file_path):
        os.remove(file_path)
        print(f"File '{file_path}' is removed.")
    else:
        print(f"File '{file_path}' not found.")
except Exception as e:
    print(f"Error: {e}")
```

## 079 - Viết chương trình để ghi thêm nội dung vào cuối file


```python
file_path = 'example.txt'
content_to_add = "\nThis is the new content added to the file."

try:
    # Open file with 'a' mode
    with open(file_path, 'a') as file:
        file.write(content_to_add)
    print(f"Content is writen to '{file_path}'.")
except FileNotFoundError:
    print(f"Error: {file_path} not found")
except Exception as e:
    print(f"Error: {e}")
```

```python
source_file_path = "source.txt"
destination_file_path = "destination.txt"

try:
    with open(source_file_path, "r") as source_file:
        content = source_file.read()

    with open(destination_file_path, "w") as destination_file:
        destination_file.write(content)
        print("Content written")
except Exception as e:
    print(f"Error: {e}")
```
